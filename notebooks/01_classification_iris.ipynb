# Clasificación de flores Iris con Regresión Logística

En este notebook se implementa un modelo de aprendizaje supervisado para resolver un problema de clasificación.

El objetivo es predecir la especie de una flor utilizando sus características físicas mediante un modelo de Regresión Logística.

```python
# Importar el conjunto de datos Iris y el modelo de Regresión Logística
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression

# Cargar el conjunto de datos Iris
data = load_iris()

# Variables de entrada: características de las flores
X = data.data

# Variable objetivo: especie de la flor
y = data.target

# Crear el modelo de clasificación
clf = LogisticRegression(max_iter=200)

# Entrenar el modelo utilizando los datos etiquetados
clf.fit(X, y)

# Realizar predicciones sobre las primeras cinco muestras
predictions = clf.predict(X[:5])

# Mostrar las clases predichas
print(predictions)
```

## Interpretación de resultados

El modelo devuelve una lista con las clases predichas para las primeras cinco flores del conjunto de datos.

Cada número representa una especie:

- **0:** Iris Setosa.
- **1:** Iris Versicolor.
- **2:** Iris Virginica.

Estas predicciones muestran cómo un modelo de clasificación aprende patrones a partir de datos etiquetados y posteriormente utiliza ese conocimiento para clasificar nuevos ejemplos.
