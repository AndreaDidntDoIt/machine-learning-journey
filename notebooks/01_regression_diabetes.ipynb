# Introducción al Aprendizaje Supervisado

En este notebook se presenta un ejemplo sencillo del flujo de trabajo de un modelo de aprendizaje supervisado utilizando Scikit-learn.

Se utilizará el conjunto de datos **Diabetes** incluido en la biblioteca para entrenar un modelo de Regresión Lineal y realizar las primeras predicciones.

```python
# Importar las librerías necesarias
from sklearn.linear_model import LinearRegression
from sklearn.datasets import load_diabetes

# Cargar el conjunto de datos Diabetes
data = load_diabetes()

# Variables de entrada (características)
X = data.data

# Variable objetivo
y = data.target

# Crear el modelo de Regresión Lineal
model = LinearRegression()

# Entrenar el modelo
model.fit(X, y)

# Realizar predicciones sobre las primeras cinco muestras
predictions = model.predict(X[:5])

# Mostrar las predicciones
print(predictions)
```

## Interpretación

En este ejemplo se observa el flujo básico de un modelo de aprendizaje supervisado:

- Se cargó un conjunto de datos.
- Se definieron las variables de entrada (`X`) y la variable objetivo (`y`).
- Se entrenó un modelo de Regresión Lineal mediante el método `fit()`.
- Se realizaron predicciones utilizando el método `predict()`.

Este ejemplo tiene como objetivo comprender el proceso general del aprendizaje supervisado. En módulos posteriores se estudiará con mayor detalle el funcionamiento de los algoritmos de regresión y clasificación.
