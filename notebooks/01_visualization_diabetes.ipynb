# Visualización de datos: Diabetes Dataset

## Objetivo

En este notebook se realiza un análisis exploratorio mediante visualización de datos.

Se analizará la relación entre el Índice de Masa Corporal (BMI) y la progresión de la enfermedad utilizando un gráfico de dispersión.

La visualización permite identificar patrones y relaciones entre variables antes de entrenar un modelo de Machine Learning.

```python
# Importar librerías necesarias
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
import pandas as pd

# Cargar conjunto de datos Diabetes
data = load_diabetes()

# Convertir características a DataFrame
X = pd.DataFrame(data.data, columns=data.feature_names)

# Variable objetivo
y = data.target

# Crear gráfico de dispersión
plt.scatter(X['bmi'], y)

# Etiquetas del gráfico
plt.xlabel('BMI')
plt.ylabel('Disease Progression')
plt.title('BMI vs Target')

# Mostrar gráfico
plt.show()
```

## Interpretación

El gráfico muestra la relación entre el Índice de Masa Corporal (BMI) y la progresión de la enfermedad.

La distribución de los puntos permite observar si existe una tendencia entre ambas variables.

Una tendencia positiva indicaría que valores mayores de BMI podrían estar relacionados con una mayor progresión de la enfermedad.

La visualización es una herramienta importante para comprender los datos antes de seleccionar y entrenar un modelo.
